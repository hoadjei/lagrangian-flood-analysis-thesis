```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
ds = xr.open_dataset("../data/CV_FTLE_2025.zarr")
ds
```

```python
ds.plot.scatter(x = "plon", y ="plat", hue = "FTLE", cmap = "Grays", edgecolor = "none", alpha = 0.5)
plt.show()
```

```python
fg = ds.plot.scatter(
    x="plon", y="plat",
    hue="FTLE",
    edgecolor="none",
    alpha=0.5,
    col="time",
    col_wrap=7, 
    s=5,  
)

for ax in fg.axs.flat:
    ax.set_aspect("equal")

plt.show()
```

```python
ds_sub = ds.isel(time=slice(0, None, 2))

fg = ds_sub.plot.scatter(
    x="plon", y="plat",
    hue="FTLE",
    edgecolor="none",
    alpha=0.5,
    col="time",
    col_wrap=7,
    s=5,
)
for ax in fg.axs.flat:
    ax.set_aspect("equal")
plt.show()
```
