```python
import copernicusmarine
import xarray as xr
```

```python
ds = copernicusmarine.open_dataset(
  dataset_id="cmems_mod_glo_phy-cur_anfc_0.083deg_P1D-m",)
ds
```

```python
ssc = ds.sel(time = slice("2025-08-01T00:00:00","2025-08-31T00:00:00")).sel(longitude = slice(-29,-20), latitude = slice(13,20)).sel(depth = 0, method = "nearest")
ssc
```

```python
ssc.to_netcdf("ssc.nc")
```

```python
flowfield_env = xr.open_dataset("ssc.nc").squeeze()
flowfield_env
```
